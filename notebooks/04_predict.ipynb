```python
import pandas as pd
import joblib
```

```python
model = joblib.load('../models/spotify_model.pkl')

print("Model loaded!")
```

```python
sample = pd.DataFrame({
    'popularity': [70],
    'duration_ms': [210000],
    'danceability': [0.8],
    'energy': [0.7],
    'key': [5],
    'loudness': [-5.0],
    'mode': [1],
    'speechiness': [0.05],
    'acousticness': [0.1],
    'instrumentalness': [0.0],
    'liveness': [0.12],
    'valence': [0.6],
    'tempo': [120]
})

print(sample)
```

```python
prediction = model.predict(sample)

print("Predicted genre:", prediction[0])
```
